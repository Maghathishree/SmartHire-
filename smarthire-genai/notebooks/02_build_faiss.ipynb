# Notebook 02: Building and Querying the FAISS Vector Database
### SmartHire GenAI — Capstone Experimentation
This notebook demonstrates:
1. Loading and cleaning the Kaggle/sample job postings CSV
2. Generating dense embeddings and indexing them into a FAISS vector index (`IndexFlatL2`)
3. Serializing and persisting the index to `vectorstore/jobs_index`
4. Executing top-k semantic search queries with candidate profiles

```python
import pandas as pd
from pathlib import Path
from src.search.job_search import JobSearchEngine, load_jobs_dataframe

df = load_jobs_dataframe()
print(f"Loaded {len(df)} jobs from dataset:")
df[["job_title", "company", "location", "skills"]].head()
```

```python
# Build or reload the FAISS index
engine = JobSearchEngine()
print("Building FAISS vector database...")
engine.build_index(force_rebuild=True)
print("FAISS vector database saved to vectorstore/jobs_index!")
```

```python
# Query with a sample candidate profile
candidate_profile = """
Target Role: Data Analyst
Skills: SQL, PostgreSQL, Tableau, Excel, Power BI, Python Pandas, Data Visualization
Experience: Built executive KPI dashboards and analyzed sales transactional data
"""

matches = engine.search_jobs(candidate_profile, top_n=3)
print("Top 3 Matching Roles:")
for m in matches:
    print(f"-> {m['job_title']} at {m['company']} | Score: {m['match_score']}% (Distance: {m['raw_distance']})")
```
