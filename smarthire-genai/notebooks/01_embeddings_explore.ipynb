# Notebook 01: Embeddings Exploration & Semantic Similarity
### SmartHire GenAI — Capstone Experimentation
This notebook demonstrates:
1. Generating dense vector embeddings with Google Gemini (`text-embedding-004`)
2. Computing cosine similarity between resume snippets and job postings
3. Visualizing how semantic matching outperforms traditional keyword matching

```python
import os
import numpy as np
import pandas as pd
from dotenv import load_dotenv

load_dotenv("../.env")
print("Gemini API Key configured:", bool(os.getenv("GEMINI_API_KEY")))
```

```python
# Define sample resume snippet vs two distinct job postings
resume_snippet = "Proficient in Python, building asynchronous microservices with FastAPI, and relational PostgreSQL databases."

job_backend = "Seeking a Backend Developer experienced in building scalable REST APIs in Python and SQL query optimization."
job_marketing = "Looking for an SEO specialist to run Google Ads campaigns, create blog copy, and manage social media channels."

print("Resume:", resume_snippet)
print("Job A (Backend):", job_backend)
print("Job B (Marketing):", job_marketing)
```

```python
from src.search.embed import generate_text_embedding

vec_resume = np.array(generate_text_embedding(resume_snippet))
vec_backend = np.array(generate_text_embedding(job_backend))
vec_marketing = np.array(generate_text_embedding(job_marketing))

def cosine_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

sim_backend = cosine_sim(vec_resume, vec_backend)
sim_marketing = cosine_sim(vec_resume, vec_marketing)

print(f"Cosine Similarity (Resume vs Backend Job): {sim_backend:.4f} (High Affinity)")
print(f"Cosine Similarity (Resume vs Marketing Job): {sim_marketing:.4f} (Low Affinity)")
assert sim_backend > sim_marketing, "Backend job should have higher semantic similarity than marketing job!"
```
